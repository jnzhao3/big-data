# Dataset Exploration: Wikidata Musicians

Each task folder has a `source.csv`, a `target.csv`, and a `groundtruth.csv` mapping source columns to target columns.

```python
from pathlib import Path

import pandas as pd
from IPython.display import display

DATA_DIR = Path("../datasets/WikidataHumanCurated/Musicians")
```

```python
tasks = {}
for task_dir in sorted(DATA_DIR.iterdir()):
    if task_dir.is_dir():
        tasks[task_dir.name] = {
            name: pd.read_csv(task_dir / f"{name}.csv")
            for name in ["source", "target", "groundtruth"]
        }

for task, dfs in tasks.items():
    print(task, {name: df.shape for name, df in dfs.items()})
```
